```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

math_marks = np.random.randint(35, 100, 100)
science_marks = np.random.randint(35, 100, 100)
english_marks = np.random.randint(35, 100, 100)

print(math_marks[:10])
```

```python
print("Math Average:", np.mean(math_marks))
print("Science Average:", np.mean(science_marks))
print("English Average:", np.mean(english_marks))

print("Highest Math Score:", np.max(math_marks))
print("Lowest Math Score:", np.min(math_marks))
```

```python
passing_marks = 40

pass_count = np.sum(math_marks >= passing_marks)
fail_count = np.sum(math_marks < passing_marks)

print("Students Passed (Math):", pass_count)
print("Students Failed (Math):", fail_count)
```

```python
grade_A = np.sum(math_marks >= 80)
grade_B = np.sum((math_marks >= 60) & (math_marks < 80))
grade_C = np.sum((math_marks >= 40) & (math_marks < 60))
grade_D = np.sum(math_marks < 40)

print("Grade A (80+):", grade_A)
print("Grade B (60-79):", grade_B)
print("Grade C (40-59):", grade_C)
print("Grade D (Below 40):", grade_D)
```

```python
total_marks = math_marks + science_marks + english_marks

overall_topper_index = np.argmax(total_marks)
print("Overall Topper (Student Index):", overall_topper_index)
print("Topper's Total Marks:", total_marks[overall_topper_index])

print("Math Std Dev:", np.std(math_marks))
print("Science Std Dev:", np.std(science_marks))
print("English Std Dev:", np.std(english_marks))
```

```python
subjects = ['Math', 'Science', 'English']
averages = [np.mean(math_marks), np.mean(science_marks), np.mean(english_marks)]

plt.bar(subjects, averages, color=['skyblue', 'lightgreen', 'salmon'])
plt.title('Average Marks by Subject')
plt.ylabel('Average Marks')
plt.savefig('chart1_subject_average.png')
plt.show()
```

```python
grades = ['A (80+)', 'B (60-79)', 'C (40-59)', 'D (<40)']
counts = [grade_A, grade_B, grade_C, grade_D]

plt.pie(counts, labels=grades, autopct='%1.1f%%', colors=['gold','lightblue','lightgreen','salmon'])
plt.title('Grade Distribution (Math)')
plt.savefig('chart2_grade_distribution.png')
plt.show()
```

## Key Insights:

- English had the highest average score (68.48) among the three subjects, followed by Math (67.8) and Science (65.74)
- 87% of students passed in Math (passing marks: 40), while 13% failed
- 36% of students scored Grade A (80+) in Math, showing strong overall performance
- Student #55 was the overall topper with 279 total marks across all three subjects
- English showed the most consistent scores (lowest standard deviation: 17.29), while Math had the most variation (20.96)
